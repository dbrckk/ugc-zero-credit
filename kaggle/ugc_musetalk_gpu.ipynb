```python
import subprocess
from pathlib import Path
source_dir=Path('/kaggle/temp/ugc-zero-credit-src')
if not source_dir.exists():
    subprocess.run(['git','clone','--quiet','--depth','1','--branch','main',
                    'https://github.com/dbrckk/ugc-zero-credit.git',str(source_dir)],check=True)
else:
    assert (source_dir/'.git').is_dir(), 'Source code existant non Git : refuser d’écraser.'
    subprocess.run(['git','-C',str(source_dir),'pull','--ff-only','origin','main'],check=True)
print('Code du projet:', source_dir)
```
